```python
import os
from PIL import Image
import numpy as np

# 图片文件夹路径
folder_path = '/home/lab2/2T/CredSplatting/datasets/tnt/Train/images'
folder_path = '/home/lab2/2T/CredSplatting/datasets/nerf/chair/train'

# 初始化一个列表来存储所有图片的像素值
all_pixels = []

# 遍历文件夹中的所有文件
for filename in os.listdir(folder_path):
    if filename.endswith('.jpg') or filename.endswith('.png'):  # 根据你的图片格式调整
        # 打开图片并转换为RGB格式
        img = Image.open(os.path.join(folder_path, filename)).convert('RGB')
        # 将图片转换为numpy数组并添加到列表中
        all_pixels.append(np.array(img))

# 将所有图片的像素值堆叠成一个大的numpy数组
all_pixels = np.concatenate(all_pixels, axis=0)

# 计算均值和方差
mean = np.mean(all_pixels, axis=(0, 1))
variance = np.var(all_pixels, axis=(0, 1))

print("均值:", mean)
print("方差:", variance)
```
